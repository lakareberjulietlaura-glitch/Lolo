# ERROR HANDLING.

```python
try:
    a = int(input("First number: "))
    b = int(input("Second number: "))
    print("Result:", a/b)
except ValueError:
    print("Please enter valid numbers.")
except ZeroDivisionError:
    print("Cannot divide by zero.")        
```

```python
while True:
    try:
        age  = int(input("Enter your age: "))
        break            # leave the loop if conversion worked
    except ValueError:
        print("Please enter a whole number.")

print("Age recorded:", age)        
```

 # else.

```python
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Error")
else:
    print("Success:", result)        
```

# finally

```python
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Error")
finally:
    print("Always runs.")        
```

# 1. safe division.
ask for two numbers and print their ratio. if the second is zero, show a friendly message.

```python
try: 
    c = int(input("Enter first number: "))
    d = int(input("Enter second number"))
    print("result:", c / d)
except ValueError:
    print("Please enter valid numbers.")
except ZeroDivisionError:
    print("Cannot divide by zzero.")        
```

# 2. valid integer.
Keep asking for an integer until the user types a valid one, then print.


```python
while True:
    try:
        number = int(input("Enter an integer: "))
        print("You entered:", number)
        break
    except ValueError:
        print("Invalid input. please enter an integer.")       
```

# 3. Safe_int function
write safe_int(prompt) that shows the prompt, reads input, and returns an integer. Retry on error.

```python
def safe_int(prompt):
    while True:
        try:
            return int(input(prompt))
        except ValueError:
            print("Invalid input. Please enter an integer.")
number = safe_int("Enter an integer: ")
print("You entered:", number)            
```
