```python
def safe_int(prompt):
    while True:
        try:
            return int(input(prompt))
        except ValueError:
            print("Please enter a whole number. ")

marks = []
n = safe_int("How many students? ")
for i in range(n):
    marks.append(safe_int("Mark: "))

print("Class average:", sum(marks)/ len(marks))                
```

### Task 4.
Write a function that takes a list of numbers and returns how many are above 50. Combine functions and list.

```python
def count_above_50(numbers):
    count = 0
    for number in numbers:
        if number > 50:
            count+= 1
    return count

numbers = [30, 40, 55, 77, 90, 20, 65, 88]
result = count_above_50(numbers)
print("Numbers above 50:", result)        
```

### Task 5 

Write a small menu (Add/view/Exit) using if elif optional: try match case on 3.10+. Use a while true menu for the menu.

```python
while True:
    print("\n1. Add items")
    print("2. View items")
    print("3. Exit")
    choice = input("Choice: ")

    match choice:
        case "1":
            print("Adding items....")
        case "2":
            print("Viewing items....")
        case "3":
            print("Goodbye")
            break
        case _:
            print("Invalid option. try again.")        
```

### Task 2.
Read n marks into a list, then print average, higest and lowest. Use sum, max, min, len.

```python
marks = []

n = int(input("How many marks do you want to enter? "))

for i in range(n):
    mark = float(input("Enter mark: "))
    marks.append(mark)

average = sum(marks) / len(marks)
highest = max(marks)
lowest = min(marks)

print("Average:", average)
print("Highest:", highest)
print("Lowest:", lowest)
```

Match_case is a cleaner way to choose between many fixed options( like a stronger if_elif chain)

```python
command = input("Enter command: ")

match command:
    case "starts":
        print("Starting...")
    case "stop":
        print("stopping...")
    case "help":
        print("Available commands: start, stop, help")
    case _ :
       print("Unknown command")            
```
